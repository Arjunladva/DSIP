```python
import cv2
def perform_gray_level_operation(image, operation):
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    if operation == 'contrast':
        processed_image = cv2.equalizeHist(gray_image)
    elif operation == 'brightness':
        beta = 50
        processed_image = cv2.convertScaleAbs(gray_image, alpha=1.0, beta=beta)
    elif operation == 'thresholding':
        _, processed_image = cv2.threshold(
            gray_image, 127, 255, cv2.THRESH_BINARY
        )
    else:
        print("Invalid operation. Available operations: "
              "'contrast', 'brightness', 'thresholding'")
        return None
    processed_image = cv2.cvtColor(processed_image, cv2.COLOR_GRAY2BGR)
    return processed_image
image_path = '1.jpg'
input_image = cv2.imread(image_path)
if input_image is None:
    print("Error: Could not load the input image.")
else:
    operation_type = 'contrast'
    output_image = perform_gray_level_operation(
        input_image,
        operation_type
    )
    if output_image is not None:
        cv2.imshow('Processed Image', output_image)
        output_path = 'output_image.jpg'
        cv2.imwrite(output_path, output_image)
        print(f"Processed image saved at: {output_path}")
        cv2.waitKey(0)
        cv2.destroyAllWindows()
```
