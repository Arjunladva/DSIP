```python
import numpy as np
import matplotlib.pyplot as plt
n = np.arange(-3, 7)

x1 = []
for i in n:
    if i % 2 == 0:
        x1.append(3)
    else:
        x1.append(2)

plt.figure(figsize=(6,4))
plt.stem(n, x1, basefmt="k")
plt.title("Problem 1")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
```

```python
n2 = np.arange(0, 6)
x2 = [1,2,3,3,2,1]

plt.figure(figsize=(6,4))
plt.step(n2, x2, where='post', linewidth=2)
plt.xlim(-0.5,6)
plt.ylim(0,4)
plt.title("Problem 2")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
```

```python
n3 = np.linspace(-2,2,400)
x3 = np.piecewise(
    n3,
    [n3 < -1,
     (n3 >= -1) & (n3 <= 1),
     (n3 > 1)],
    [-2,
     lambda n: 2*n,
     2]
)

plt.figure(figsize=(6,4))
plt.plot(n3, x3, linewidth=2)
plt.xlim(-2.2,2.2)
plt.ylim(-2.5,2.5)
plt.title("Problem 3")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
```

```python
n4 = np.arange(-2, 11)
u = lambda n: (n >= 0).astype(int)
x4 = u(n4) - u(n4-3) - 5*u(n4-7)

plt.figure(figsize=(6,4))
plt.stem(n4, x4, basefmt="k")
plt.title("Problem 4")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
```

```python
n5 = np.arange(-3,4)
delta = lambda n: (n == 0).astype(int)
x5 = delta(n5) + 3*delta(n5-1) + 5*delta(n5+1)

plt.figure(figsize=(6,4))
plt.stem(n5, x5, basefmt="k")
plt.title("Problem 5")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
plt.show()
```
